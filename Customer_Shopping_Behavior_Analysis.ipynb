```python

import pandas as pd

df = pd.read_csv('customer_shopping_behavior.csv')
df.head()
```

```python
# describe the statistics of numeric column 
df.describe()
```

```python
#  Describe the statistics of all column including categorical column
df.describe(include='all')
```

```python
# check missing values
df.isnull().sum()
```

```python
# Count available (non-null) review ratings by category
df.groupby('Category')['Review Rating'].count()
```

```python
# fill the missing values with the median of each category.
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform('median')
```

```python
df.isnull().sum()
```

```python
# Transform the columns name in lower case
df.columns = df.columns.str.lower()

# Replace the spaces with undercorse
df.columns = df.columns.str.replace(' ','_')

df = df.rename(columns = {'purchase_amount_(usd)' : 'purchase_amount'})

df.columns
```

```python
# create a column age group
labels = ['Young Adult', 'Adult', 'Middle Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)

df[['age', 'age_group']].head(30)
```

```python
# create column purchase_frequency_days

frequency_mapping={
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly': 30,
    'Yearly' : 365,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Every 3 Months' : 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```
